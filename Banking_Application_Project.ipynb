```python
import requests
import gspread
from oauth2client.service_account import ServiceAccountCredentials

# --- 1. Google Sheets Setup ---
# 'service_account.json' wo file hai jo aap Google Cloud se download karenge
scope = ["https://spreadsheets.google.com/feeds", "https://www.googleapis.com/auth/drive"]
creds = ServiceAccountCredentials.from_json_keyfile_name("service_account.json", scope)
client = gspread.authorize(creds)

# Spreadsheet open karein
sheet_url = "https://docs.google.com/spreadsheets/d/1XgffcxO4ePjnp-91XlA6FEHdZKZWvR5yxAghgMerbWo/edit#gid=0"
sheet = client.open_by_url(sheet_url).sheet1 # Pehli sheet select ki

# Pehle se maujood links nikalna (taaki duplicates na ho)
existing_links = sheet.col_values(3) # Maan lete hain Column 3 mein links hain

# --- 2. GitHub Details ---
owner = "AamirAli786-pro"
repo = "Portfolio_Image"
api_url = f"https://api.github.com/repos/{owner}/{repo}/contents/"

response = requests.get(api_url)

if response.status_code == 200:
    files = response.json()
    new_rows = []

    print("Checking for new images...")

    for file in files:
        if file['name'].lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.svg')):
            permalink = file['download_url']

            # CHECK: Agar link pehle se GSheet mein nahi hai toh hi add karein
            if permalink not in existing_links:
                commit_url = f"https://api.github.com/repos/{owner}/{repo}/commits?path={file['path']}&per_page=1"
                commit_resp = requests.get(commit_url).json()

                if commit_resp:
                    upload_date = commit_resp[0]['commit']['committer']['date']
                    new_rows.append([file['name'], upload_date, permalink])
                    print(f"New Found: {file['name']}")

    # --- 3. GSheet mein Update karna ---
    if new_rows:
        sheet.append_rows(new_rows)
        print(f"Successfully added {len(new_rows)} new images to GSheet!")
    else:
        print("Koi naya image nahi mila.")

else:
    print("Error: GitHub data fetch nahi ho paya.")
```
